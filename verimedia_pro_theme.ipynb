```python
import subprocess
import time
from pyngrok import ngrok

print("\n═" * 60)
print("[VERIMEDIA] Starting Live Server")
print("═" * 60)

# Kill existing processes
print("\n[→] Cleaning up previous processes...")
os.system('lsof -ti:5173 | xargs kill -9 2>/dev/null; true')
time.sleep(1)

# Start server
print("[→] Starting Vite development server...")
process = subprocess.Popen(
    ['npm', 'run', 'dev'],
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE,
    text=True
)

print("[→] Waiting for server initialization...")
time.sleep(4)

# Create tunnel
try:
    print("[→] Creating ngrok tunnel...")
    public_url = ngrok.connect(5173, "http")
    
    print("\n" + "═" * 60)
    print("[✓] VERIMEDIA IS LIVE")
    print("═" * 60)
    print(f"\n🌐 PUBLIC URL: {public_url}")
    print("\n📱 CLICK THE LINK ABOVE TO ACCESS VERIMEDIA")
    print("\n⚠️  KEEP THIS NOTEBOOK RUNNING WHILE USING THE APP")
    print("\n" + "═" * 60 + "\n")

except Exception as e:
    print(f"\n[✗] Error: {e}")
    print("[→] Retrying in 2 seconds...")
```

<hr class="section-divider">

<div class="info-box">
  ℹ️ <strong>Server Monitor</strong><br>
  The cell below monitors the server status. Keep it running to maintain your connection.
</div>

```python
import time

try:
    counter = 0
    while True:
        time.sleep(30)
        counter += 1
        
        if process.poll() is not None:
            print(f"\n[✗] Server stopped at {counter * 30}s")
            break
        else:
            elapsed = counter * 30
            status = "[●]" if counter % 2 == 0 else "[○]"
            print(f"{status} Server running - {elapsed}s elapsed")

except KeyboardInterrupt:
    print("\n[→] Shutting down...")
    process.terminate()
    ngrok.kill()
    print("[✓] Server stopped cleanly")
```
